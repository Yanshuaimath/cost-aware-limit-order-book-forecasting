# 07 — Walk-forward results

This notebook aggregates out-of-sample fold metrics created by `scripts/run_walk_forward.py`.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

RESULTS = Path('../outputs/btc_h40_cost/xgboost_walk_forward/walk_forward_metrics.csv')
RESULTS
```

```python
if not RESULTS.exists():
    print('Run create_walk_forward_folds.py and run_walk_forward.py first.')
    results = None
else:
    results = pd.read_csv(RESULTS)
    display(results)
```

```python
if results is not None:
    summary_columns = [column for column in ['macro_f1', 'mcc', 'log_loss', 'ece_10', 'backtest_total_return', 'backtest_max_drawdown', 'backtest_n_trades'] if column in results]
    display(results[summary_columns].agg(['mean', 'std', 'min', 'max']).T)
```

```python
if results is not None and 'macro_f1' in results:
    ax = results.plot(x='fold', y='macro_f1', marker='o', title='Out-of-sample macro-F1 by fold', figsize=(9, 4))
    ax.set_ylim(0, 1)
    plt.show()
```

```python
if results is not None and 'backtest_total_return' in results:
    ax = results.plot(x='fold', y='backtest_total_return', kind='bar', title='Net portfolio return by test fold', figsize=(9, 4))
    ax.axhline(0)
    plt.show()
```

## Interpretation

Report the dispersion across folds, not only the average. A model that performs well on one day and fails on another may not be robust enough for a resume claim.
