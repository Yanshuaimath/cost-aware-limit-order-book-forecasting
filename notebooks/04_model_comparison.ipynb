## Compare held-out performance

Load each run’s saved test metrics and rank the models by macro-F1. Higher is better for macro-F1, MCC, balanced accuracy, and samples/second; lower is better for ECE.

```python
RUNS = [
    REPO_ROOT / "outputs/synthetic_baseline_h50",
    REPO_ROOT / "outputs/synthetic_deeplob_h50",
    REPO_ROOT / "outputs/synthetic_transformer_h50",
]
rows = []
for run in RUNS:
    metrics_path = run / "test_metrics.json"
    if metrics_path.exists():
        metrics = json.loads(metrics_path.read_text())
        rows.append({
            "run": run.name,
            "macro_f1": metrics["macro_f1"],
            "mcc": metrics["mcc"],
            "balanced_accuracy": metrics["balanced_accuracy"],
            "ece_10": metrics["ece_10"],
            "samples_per_second": metrics["samples_per_second"],
        })
pd.DataFrame(rows).sort_values("macro_f1", ascending=False) if rows else "Evaluate models first."
```
