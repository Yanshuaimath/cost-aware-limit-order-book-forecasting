# 04 - Model Comparison

```python

from pathlib import Path
import json, sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))
PROCESSED = ROOT / "data/processed/synthetic_h50"
print("Repository:", ROOT)
print("Processed data:", PROCESSED)
```

```python
RUNS = [
    ROOT / "outputs/baseline_h50",
    ROOT / "outputs/deeplob_h50",
    ROOT / "outputs/transformer_h50",
]
rows = []
for run in RUNS:
    metrics_path = run / "test_metrics.json"
    if metrics_path.exists():
        metrics = json.loads(metrics_path.read_text())
        rows.append({
            "run": run.name,
            "macro_f1": metrics["macro_f1"],
            "mcc": metrics["mcc"],
            "balanced_accuracy": metrics["balanced_accuracy"],
            "ece_10": metrics["ece_10"],
            "samples_per_second": metrics["samples_per_second"],
        })
pd.DataFrame(rows).sort_values("macro_f1", ascending=False) if rows else "Evaluate models first."
```

Model ranking should be discussed jointly across predictive quality, calibration, latency, and cost-aware diagnostics—not macro-F1 alone.
